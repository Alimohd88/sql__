````python
%%sql
select s.orderdate,s.quantity * s.netprice * s.exchangerate as  NETREVENUE,c.givenname,c.surname,c.countryfull,c.continent,p.productkey,p.productname,p.categoryname,p.subcategoryname,
CASE
  WHEN s.quantity * s.netprice * s.exchangerate >1000
  THEN 'HIGH' ELSE 'LOW'
END AS LOW_HIGH_ALIAS
from sales s
left  join customer c
on s.customerkey=c.customerkey
left join product p
on s.productkey=p.productkey
where orderdate::date >='2020-01-01'

limit 10
````

````python
%%sql

SELECT
    c.countryfull,
    p.productname,
    p.categoryname,

    SUM(s.quantity * s.netprice * s.exchangerate) AS NETREVENUE,

    CASE
        WHEN SUM(s.quantity * s.netprice * s.exchangerate) > 1000
        THEN 'HIGH'
        ELSE 'LOW'
    END AS LOW_HIGH_ALIAS

FROM sales s

LEFT JOIN customer c
    ON s.customerkey = c.customerkey

LEFT JOIN product p
    ON s.productkey = p.productkey

WHERE s.orderdate::date >= '2020-01-01'

GROUP BY
    c.countryfull,
    p.productname,
    p.categoryname

LIMIT 10;
````

````python
%%sql
SELECT s.orderdate,count(distinct(s.customerkey)) as totalcustomer,
count(DISTINCT
  CASE
    WHEN c.continent = 'Europe' then s.customerkey end) as eu_customers,
  count(distinct case when c.continent='North America' then c.customerkey end ) as na_customers,
  count(distinct case when c.continent='Australia' then c.customerkey end ) as as_customers
from sales s

left  join customer c
on s.customerkey=c.customerkey
where s.orderdate::date between '2023-01-01' and '2023-12-31'
group by s.orderdate
order by s.orderdate
````

````python
%%sql
SELECT
    p.categoryname,
    SUM(CASE WHEN EXTRACT(YEAR FROM s.orderdate) = 2022 THEN s.quantity * s.netprice * s.exchangerate ELSE 0 END) AS total_revenue_2022,
    SUM(CASE WHEN EXTRACT(YEAR FROM s.orderdate) = 2023 THEN s.quantity * s.netprice * s.exchangerate ELSE 0 END) AS total_revenue_2023
FROM sales s
LEFT JOIN product p
    ON s.productkey = p.productkey
WHERE EXTRACT(YEAR FROM s.orderdate) IN (2022, 2023)
GROUP BY p.categoryname
ORDER BY p.categoryname;
````

````python
%%sql
select orderdate,quantity,netprice,
case
  when quantity >=2 and netprice >=100 then 'multiple  high value  items'
  when netprice >=100 then 'sinngle high value item'
  when quantity >=2 then 'multiple standard item'
  else 'single standard item'
end as size
from sales

order by quantity desc

````

````python
%%sql

WITH sales_segment AS (

    SELECT
        quantity,
        CASE
            WHEN quantity >= 2 AND netprice >= 100
                THEN 'multiple high value items'

            WHEN netprice >= 100
                THEN 'single high value item'

            WHEN quantity >= 2
                THEN 'multiple standard item'

            ELSE 'single standard item'
        END AS size

    FROM sales
)

SELECT
    quantity,
    size,
    COUNT(*) AS total_count

FROM sales_segment

GROUP BY
    quantity,
    size

ORDER BY
    quantity desc,
    size desc ;
````

````python
%%sql
with median_value  as(
select PERCENTILE_CONT(0.5) WITHIN GROUP (order by (s.quantity*s.netprice*s.exchangerate)) as md
from sales s
where orderdate between '2022-01-01' and '2022-12-31'
)


SELECT
    p.categoryname AS category,
  sum(case when (s.quantity*s.netprice*s.exchangerate)< md
  and orderdate between '2022-01-01' and '2022-12-31'
   then (s.quantity*s.netprice*s.exchangerate) end) as low_net_rev_2022,

   sum(case when (s.quantity*s.netprice*s.exchangerate)>= md
   and  orderdate between '2022-01-01' and '2022-12-31'
   then (s.quantity*s.netprice*s.exchangerate) end) as high_net_rev_2022,

    sum(case when (s.quantity*s.netprice*s.exchangerate)< md
  and orderdate between '2023-01-01' and '2023-12-31'
   then (s.quantity*s.netprice*s.exchangerate) end) as low_net_rev_2023,

   sum(case when (s.quantity*s.netprice*s.exchangerate)>= md
   and  orderdate between '2023-01-01' and '2023-12-31'
   then (s.quantity*s.netprice*s.exchangerate) end) as high_net_rev_2023



FROM sales s

LEFT JOIN product p
    ON s.productkey = p.productkey, median_value  mv

GROUP BY
    p.categoryname

ORDER BY
    p.categoryname;
````

````python
%%sql

WITH percentile AS (

    SELECT
        PERCENTILE_CONT(0.25) WITHIN GROUP (
            ORDER BY (s.quantity * s.netprice * s.exchangerate)
        ) AS reve_25TH_percentile,

        PERCENTILE_CONT(0.75) WITHIN GROUP (
            ORDER BY (s.quantity * s.netprice * s.exchangerate)
        ) AS reve_75TH_percentile

    FROM sales s

    WHERE orderdate BETWEEN '2022-01-01' AND '2022-12-31'
)

SELECT
    p.categoryname AS category,
    case
      when (s.quantity * s.netprice * s.exchangerate) <= perc.reve_25TH_percentile then '3-low'
      when (s.quantity * s.netprice * s.exchangerate) >= perc.reve_75TH_percentile then '1-high'
      else '2-medium'
    end as revenue_tier,
    SUM(s.quantity * s.netprice * s.exchangerate) as total_revenue

FROM sales s

LEFT JOIN product p
    ON s.productkey = p.productkey,

percentile perc

GROUP BY p.categoryname,revenue_tier

ORDER BY category,revenue_tier;
````
